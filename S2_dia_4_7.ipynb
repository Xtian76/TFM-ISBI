```python
# Step 9: Visualize clustering results (Hierarchical Clustering with cut dendrogram)

from scipy.cluster.hierarchy import fcluster
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

print("Obteniendo etiquetas de cluster y visualizando resultados del clustering jerárquico...")

# Define the number of clusters to visualize for hierarchical clustering
k_values_to_visualize = [2, 3]

for metric, dist_matrices_results in hierarchical_results.items():
    for dist_metric, linkage_results in dist_matrices_results.items():
        print(f"\nVisualizando resultados para '{metric}' usando '{dist_metric}' distance:")
        for linkage_type, linked_matrix in linkage_results.items():
            print(f"  Linkage: {linkage_type}")

            for k in k_values_to_visualize:
                print(f"    Obteniendo etiquetas para k = {k}")
                try:
                    # Get cluster labels by cutting the dendrogram at maxclust k
                    cluster_labels = fcluster(linked_matrix, k, criterion='maxclust')

                    # Add labels to a DataFrame for easier visualization/analysis
                    labels_df = pd.DataFrame({
                        'Sample': distance_matrices_loaded[metric][dist_metric].index.tolist(),
                        'Cluster': cluster_labels
                    })

                    print(f"      Etiquetas obtenidas para k = {k}.")
                    # print(labels_df.head()) # Uncomment to see the labels dataframe

                    # --- Visualization ---
                    # For visualization, it's helpful to use a dimensionality reduction technique
                    # to project the distance matrix into 2D or 3D space.
                    # MDS (Multidimensional Scaling) is suitable for distance matrices.
                    from sklearn.manifold import MDS

                    print("      Aplicando MDS para visualización...")
                    try:
                        # Ensure the distance matrix is in the correct format for MDS (symmetric, positive)
                        # Since we already converted to condensed form for linkage, we can convert back to square if needed for MDS
                        # Or, better, apply MDS directly to the original square distance matrix if it's proper distances.
                        # Assuming the loaded distance matrices are proper distance matrices (symmetric, non-negative, zero diagonal).
                        mds = MDS(n_components=2, dissimilarity='precomputed', random_state=42, normalized_stress='auto')
                        mds_coords = mds.fit_transform(distance_matrices_loaded[metric][dist_metric])
                        print("      MDS completado.")

                        # Create a DataFrame for MDS coordinates and add cluster labels
                        mds_df = pd.DataFrame(mds_coords, columns=['MDS1', 'MDS2'])
                        mds_df['Cluster'] = cluster_labels
                        mds_df['Sample'] = distance_matrices_loaded[metric][dist_metric].index.tolist() # Add sample names

                        # Plot the clusters in the MDS reduced space
                        plt.figure(figsize=(10, 7))
                        sns.scatterplot(data=mds_df, x='MDS1', y='MDS2', hue='Cluster', palette='viridis', s=100)

                        # Optional: Annotate points with sample names (can be cluttered for many points)
                        # for i in range(len(mds_df)):
                        #     plt.text(mds_df['MDS1'][i], mds_df['MDS2'][i], mds_df['Sample'][i], fontsize=9)

                        plt.title(f"Visualización de Clusters Jerárquicos (k={k}) para {metric} ({dist_metric} - {linkage_type}) (MDS)")
                        plt.xlabel("MDS1")
                        plt.ylabel("MDS2")
                        plt.grid(True)
                        plt.show()

                    except Exception as e:
                        print(f"      Error durante la aplicación de MDS o la visualización: {e}")

                except Exception as e:
                    print(f"    Error obteniendo etiquetas para k = {k}: {e}")

print("\nVisualización de resultados del clustering jerárquico completa.")
```

**Resumen Final del Análisis de Clustering de Fenotipos**

Este análisis tuvo como objetivo identificar subgrupos (fenotipos) dentro de los datos de series de tiempo utilizando varias métricas y algoritmos de clustering. Se utilizaron matrices de distancia precalculadas (Euclidiana y DTW) y se calculó una matriz de distancia de similitud del coseno.

**Hallazgos Clave:**

1.  **Matrices de Distancia Originales (Euclidiana y DTW):** La verificación reveló que estas matrices contenían predominantemente valores cero en el triángulo superior (excluyendo la diagonal). Esto indica que, según las métricas y el cálculo utilizados para generar estas matrices, todas las series de tiempo eran idénticas o tenían distancia cero entre sí. Como resultado, los algoritmos de clustering (K-means, GMM, DBSCAN, Jerárquico) aplicados a estas matrices solo identificaron un **único cluster** que contenía todas las muestras. Esto hizo imposible evaluar el clustering con métricas de validación interna para múltiples clusters o identificar agrupaciones distintas. La causa raíz de estas matrices con ceros requiere una investigación de los datos originales o el proceso de cálculo de distancia inicial.
2.  **Matriz de Distancia de Similitud del Coseno:** Se calculó una matriz de distancia utilizando la similitud del coseno (1 - similitud). La inspección inicial reveló valores mínimos y máximos de similitud dentro del rango [-1, 1], pero la matriz de distancia derivada contenía un pequeño número de valores negativos (probablemente debido a imprecisiones de punto flotante) y la diagonal no era exactamente cero.
3.  **Corrección y Clustering con Distancia de Similitud del Coseno:** Se corrigieron los valores negativos en la matriz de distancia de similitud del coseno estableciéndolos a cero. Luego se aplicaron los algoritmos de clustering a estas matrices corregidas:
    *   **DBSCAN:** Logró identificar **múltiples clusters** y outliers para cada métrica, lo que sugiere la existencia de estructuras de densidad basadas en la distancia de similitud del coseno.
    *   **K-means y GMM:** Se aplicaron con k=2 a 5 y se calcularon métricas de validación interna (Silhouette, Davies-Bouldin, Calinski-Harabasz). Los resultados de estas métricas y las curvas del método del codo (`inertia_results_cosine`) proporcionaron información para ayudar en la selección del número óptimo de clusters (tarea que requiere análisis e interpretación manual o asistida).
    *   **Clustering Jerárquico:** Tuvo problemas técnicos con la conversión de la matriz de distancia corregida a formato condensado debido a que la diagonal no era exactamente cero.
4.  **Visualización:** Se visualizó la estructura de clustering para la distancia de similitud del coseno corregida utilizando MDS para reducir la dimensionalidad a 2D y graficando los puntos coloreados por las etiquetas de cluster (utilizando un número de clusters de ejemplo, como k=3 de K-means o k=2 de GMM en la visualización final). Estas visualizaciones mostraron la separación de las muestras en el espacio reducido según las agrupaciones encontradas.

**Conclusión:**

El análisis de clustering con las matrices de distancia Euclidiana y DTW originales **no reveló múltiples fenotipos distintos**, muy probablemente debido a que esas matrices indicaban distancias cero entre todas las muestras, lo que apunta a un problema en la preparación inicial de esos datos.

El análisis utilizando la **distancia de similitud del coseno (corregida)** sí mostró evidencia de **múltiples agrupaciones** a través de DBSCAN, K-means y GMM. **Paso 5: Cálculo de Tamaños del Efecto y Consideraciones para la Interpretación**

Hemos realizado tests estadísticos para comparar las características estáticas entre los fenotipos identificados. Estos tests (Kruskal-Wallis para numéricas, Chi-squared para categóricas) nos dan un **p-value** que nos ayuda a determinar si las diferencias observadas son estadísticamente significativas.

*   Un p-value bajo (típicamente < 0.05) sugiere que es poco probable que las diferencias observadas se deban solo al azar, indicando que hay una diferencia real entre los grupos.
*   Un p-value alto sugiere que no hay suficiente evidencia para concluir que existe una diferencia significativa.

Sin embargo, la significación estadística no es lo mismo que la significación práctica o clínica. Aquí es donde entra el **tamaño del efecto**.

**Tamaño del Efecto:**

El tamaño del efecto es una medida estandarizada de la magnitud de la diferencia entre grupos o de la fuerza de una asociación. Diferentes tests requieren diferentes medidas de tamaño del efecto:

*   **Para comparaciones numéricas entre múltiples grupos (como después de Kruskal-Wallis):** Medidas como la eta cuadrada (η²) o épsilon cuadrada (ε²) se utilizan a menudo para estimar la proporción de la varianza en la característica numérica que es explicada por la pertenencia al grupo (fenotipo). Interpretación común: pequeño (0.01), mediano (0.06), grande (0.14). Sin embargo, estas medidas globales no nos dicen qué pares de grupos específicos difieren. Para eso, necesitaríamos realizar tests post-hoc (como Dunn's test) y calcular tamaños del efecto para comparaciones pareadas (e.g., diferencias de medias estandarizadas como el Cohen's d o, más apropiado para pruebas no paramétricas, la diferencia de rangos media estandarizada).
*   **Para asociaciones categóricas (después de Chi-squared):** Medidas como V de Cramer (para tablas de contingencia mayores a 2x2) o Phi (Φ) (para tablas 2x2) se utilizan para medir la fuerza de la asociación entre la característica categórica y la pertenencia al fenotipo. Interpretación común para V de Cramer: pequeño (0.1), mediano (0.3), grande (0.5).

**Cálculo del Tamaño del Efecto (Ejemplos):**

El cálculo exacto del tamaño del efecto puede ser complejo y a menudo requiere librerías adicionales (como `pingouin` para Cohen's d después de tests post-hoc, o cálculos manuales/con funciones específicas para V de Cramer).

*   **Ejemplo conceptual para Cohen's d (comparación pareada numérica):** $d = (\text{media}_1 - \text{media}_2) / \text{desviación estándar combinada}$
*   **Ejemplo conceptual para V de Cramer (asociación categórica):** $V = \sqrt{\chi^2 / (n \times \min(k-1, r-1))}$, donde $\chi^2$ es el estadístico Chi-squared, $n$ es el tamaño total de la muestra, $k$ es el número de columnas y $r$ es el número de filas en la tabla de contingencia.

**Interpretación de los Resultados:**

Al interpretar los resultados del Paso 4 y considerar el tamaño del efecto (Paso 5), debes enfocarte en:

*   **Características con p-value bajo:** Estas son las características que probablemente difieren entre los fenotipos.
*   **Tamaño del efecto asociado:** Para las características significativas, ¿cuán grandes son las diferencias? Un p-value bajo con un tamaño del efecto pequeño puede no ser tan interesante clínicamente como un p-value bajo con un tamaño del efecto grande.
*   **Resultados Post-hoc (si se realizan):** Si se realizan tests post-hoc, estos te dirán *qué pares específicos* de fenotipos son diferentes para una característica dada.

```python
# Step 7: Perform Elbow method for K-means
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

print("\nPerforming Elbow method for K-means...")

# Store the inertia (within-cluster sum of squares) for each k
inertia_results = {}

for metric, distance_matrices in distance_matrices_loaded.items():
    inertia_results[metric] = {}
    for dist_metric, df in distance_matrices.items():
        if df is not None:
            inertia_results[metric][dist_metric] = []
            data_for_elbow = df.values

            print(f"\n  Performing Elbow method for '{metric}' using '{dist_metric}' distance:")
            # Calculate inertia for k from 1 to 10 (or a suitable range)
            # We start from 1 here for the elbow method calculation
            for k in range(1, 11):
                try:
                    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
                    kmeans.fit(data_for_elbow)
                    inertia_results[metric][dist_metric].append(kmeans.inertia_)
                    print(f"    K-means (k={k}) inertia calculated.")
                except Exception as e:
                    print(f"    Error calculating inertia for k={k}: {e}")
                    inertia_results[metric][dist_metric].append(None) # Append None if error occurs

            # Plot the Elbow curve
            plt.figure(figsize=(8, 5))
            plt.plot(range(1, 11), inertia_results[metric][dist_metric], marker='o')
            plt.title(f'Elbow Method for {metric} ({dist_metric})')
            plt.xlabel('Number of clusters (k)')
            plt.ylabel('Inertia (Within-cluster sum of squares)')
            plt.xticks(range(1, 11))
            plt.grid(True)
            plt.show()

        else:
            print(f"\n  Skipping Elbow method for '{metric}' using '{dist_metric}' distance as data is not loaded.")

print("\nElbow method complete.")
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
import matplotlib.pyplot as plt
import numpy as np

# Step 3: Apply Hierarchical clustering (Ward, Complete linkage)
hierarchical_results = {}

for metric, distance_matrices in distance_matrices_loaded.items():
    hierarchical_results[metric] = {}
    for dist_metric, df in distance_matrices.items():
        if df is not None:
            print(f"\nApplying Hierarchical clustering for '{metric}' using '{dist_metric}' distance...")
            # Hierarchical clustering typically uses a condensed distance matrix or the original data
            # Since we have the square distance matrix, we can use it directly with linkage
            # However, linkage expects a condensed distance matrix (upper triangle) or the original observation matrix.
            # Using the square matrix might work but is not the standard input.
            # A better approach is to re-calculate the condensed distance matrix if needed, or use the original data.
            # For now, we'll convert the square matrix to a condensed distance matrix format.
            # Ensure the distance matrix is symmetric and has a zero diagonal for this conversion.

            # Check if the matrix is square and numeric
            if not df.shape[0] == df.shape[1] or not np.issubdtype(df.values.dtype, np.number):
                 print(f"  Skipping hierarchical clustering for '{metric}' using '{dist_metric}' distance: Input is not a square numeric matrix.")
                 continue

            # Convert square distance matrix to condensed format (upper triangle)
            # This assumes the diagonal is 0 and the matrix is symmetric
            try:
                from scipy.spatial.distance import squareform
                condensed_dist_matrix = squareform(df.values)
            except ValueError as e:
                print(f"  Error converting square matrix to condensed format for '{metric}' using '{dist_metric}' distance: {e}")
                print("  Skipping hierarchical clustering for this matrix.")
                continue


            # Apply Ward linkage
            print("  Applying Ward linkage...")
            try:
                linked_ward = linkage(condensed_dist_matrix, 'ward')
                hierarchical_results[metric][dist_metric] = {'ward_linkage': linked_ward}
                print("    Ward linkage completed.")

                # Optional: Plot dendrogram (can be time-consuming for large datasets)
                # plt.figure(figsize=(10, 7))
                # dendrogram(linked_ward,
                #            orientation='top',
                #            labels=df.index.tolist(),
                #            distance_sort='descending',
                #            show_leaf_counts=True)
                # plt.title(f"Dendrogram for {metric} ({dist_metric} - Ward)")
                # plt.show()

            except Exception as e:
                print(f"    Error applying Ward linkage: {e}")

            # Apply Complete linkage
            print("  Applying Complete linkage...")
            try:
                linked_complete = linkage(condensed_dist_matrix, 'complete')
                hierarchical_results[metric][dist_metric]['complete_linkage'] = linked_complete
                print("    Complete linkage completed.")

                # Optional: Plot dendrogram
                # plt.figure(figsize=(10, 7))
                # dendrogram(linked_complete,
                #            orientation='top',
                #            labels=df.index.tolist(),
                #            distance_sort='descending',
                #            show_leaf_counts=True)
                # plt.title(f"Dendrogram for {metric} ({dist_metric} - Complete)")
                # plt.show()

            except Exception as e:
                print(f"    Error applying Complete linkage: {e}")

        else:
            print(f"\nSkipping Hierarchical clustering for '{metric}' using '{dist_metric}' distance as data is not loaded.")

print("\nHierarchical clustering complete.")

# The hierarchical_results dictionary now contains the linkage matrices
# You can use fcluster to get cluster labels for a given distance threshold or number of clusters
# For example, to get clusters for k=3 using Ward linkage on total_volume euclidean distance:
# ward_labels_k3 = fcluster(hierarchical_results['total_volume']['euclidean']['ward_linkage'], 3, criterion='maxclust')
# print("\nSample Ward linkage labels for total_volume (Euclidean, k=3):", ward_labels_k3)
```

```python
from sklearn.cluster import KMeans

# Step 2: Apply K-means clustering (k = 2, 3, 4, 5)
kmeans_results = {}

for metric, distance_matrices in distance_matrices_loaded.items():
    kmeans_results[metric] = {}
    for dist_metric, df in distance_matrices.items():
        if df is not None:
            kmeans_results[metric][dist_metric] = {}
            # K-means requires a data matrix, not a distance matrix directly.
            # For now, we'll use the distance matrix as input, keeping in mind this isn't the standard way
            # and might not yield optimal results. A better approach for distance matrices is hierarchical clustering.
            # If the original data is available, it would be better to use that for K-means.
            # Assuming for now we proceed with the distance matrix as input data.
            data_for_kmeans = df.values

            print(f"\nApplying K-means for '{metric}' using '{dist_metric}' distance...")
            for k in range(2, 6):
                print(f"  Applying K-means with k = {k}")
                try:
                    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10) # Added n_init for robustness
                    kmeans_labels = kmeans.fit_predict(data_for_kmeans)
                    kmeans_results[metric][dist_metric][f'k_{k}'] = kmeans_labels
                    print(f"    K-means with k={k} completed.")
                except Exception as e:
                    print(f"    Error applying K-means with k={k}: {e}")
        else:
            print(f"\nSkipping K-means for '{metric}' using '{dist_metric}' distance as data is not loaded.")

print("\nK-means clustering complete.")

# You can now access the results like:
# total_volume_euclidean_kmeans_k3_labels = kmeans_results['total_volume']['euclidean']['k_3']
```
